```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Cesta k datasetu
dataset_cesta = '/content/dataset_cleaned'

# Načtení trénovacích dat
train_data = tf.keras.preprocessing.image_dataset_from_directory(
    dataset_cesta,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(128, 128),
    batch_size=32
)

# Načtení validačních dat
val_data = tf.keras.preprocessing.image_dataset_from_directory(
    dataset_cesta,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(128, 128),
    batch_size=32
)

# Definice modelu
model = models.Sequential([
    layers.InputLayer(shape=(128, 128, 3)),
    layers.Rescaling(1./255),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("\nZačínám trénovat model...")
history = model.fit(train_data, validation_data=val_data, epochs=10)

model.save('ronaldo_model.h5')
print("\nHotovo! Model je úspěšně uložen jako ronaldo_model.h5")
```
